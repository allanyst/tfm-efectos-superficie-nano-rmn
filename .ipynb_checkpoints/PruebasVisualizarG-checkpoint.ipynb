```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv(r"data\VariateKappa_G_vs_t.dat", sep=r'\s+')
```

```python
df.head()
```

```python
df.columns=["time","G_k0","G_k100","G_k1000","G_k200","G_k2000","G_k3000","G_k50","G_k500","G_k5000"]
```

```python
tau_values=1884928.7900032925/15**2*df.time
```

```python
plt.figure(figsize=(7.5,5))
plt.loglog(tau_values[(tau_values>=1e-2)&(tau_values<=50)], df.G_k0[(tau_values>=1e-2)&(tau_values<=50)]/df.G_k0[0], "o-", label=r"$\kappa=0.0$", color="navy", markersize=2, linewidth=0.5)
plt.loglog(tau_values[(tau_values>=1e-2)&(tau_values<=50)], df.G_k2000[(tau_values>=1e-2)&(tau_values<=50)]/df.G_k2000[0], "o-", label=r"$\kappa=2000.0\,\text{nm/s}$", color="gray", markersize=2, linewidth=0.5)
plt.loglog(tau_values[(tau_values>=1e-2)&(tau_values<=50)], df.G_k5000[(tau_values>=1e-2)&(tau_values<=50)]/df.G_k5000[0], "o-", label=r"$\kappa=5000.0\,\text{nm/s}$", color="orange", markersize=2, linewidth=0.5)
# plt.loglog(tau_values, G_values/G_values[0], "o-", label="Simulación", color="navy", markersize=4, linewidth=1.5)
# plt.loglog(tau_values, G_values/G_values[0], "o-", label="Simulación", color="navy", markersize=4, linewidth=1.5)
plt.plot(tau_values[(tau_values>=2)&(tau_values<=50)], tau_values[(tau_values>=2)&(tau_values<=50)]**-1.5, "--", label=r"Teoría $(t/T_D)^{-3/2}$", color="crimson", linewidth=1.5)
plt.plot(tau_values[(tau_values>=1e-2)&(tau_values<=0.1)], np.exp(-6*tau_values[(tau_values>=1e-2)&(tau_values<=0.1)]), "--", label=r"Teoría $\exp(-6t/T_D)$", color="green", linewidth=1.5)
# plt.xlabel("Tiempo (s)", fontsize=12)
plt.xlabel(r"$t/T_D$", fontsize=12)
plt.ylabel("G(t)/G(0)", fontsize=12)
plt.title("Función de Correlación Nano-NMR", fontsize=14)
plt.legend()
plt.grid(which="both", ls='--', alpha=0.7)
# plt.savefig("figuras/correlacion_nanoNMR.pdf", bbox_inches="tight", dpi=300)
plt.show()
plt.close()
```
