```python
import numpy as np
from scipy.special import jv, jn_zeros, j1
from scipy.linalg import eigh_tridiagonal
from scipy.integrate import simpson
from joblib import Parallel, delayed
import matplotlib.pyplot as plt

# =============================================
# Parámetros Físicos (Optimizados)
# =============================================
D0 = 2e-9           # Coeficiente de difusión libre [m²/s]
R = 1e-6            # Radio del cilindro [m]
L = 100e-9          # Longitud axial del dominio [m]
z_c = 10e-9         # Altura crítica de transición [m]
lambda_param = 0.2  # Parámetro de decaimiento [nm⁻¹]
kappa = 0.0         # Tasa de adsorción superficial [s⁻¹]
rho0 = 0.5e-6       # Posición radial inicial [m]
z0 = 5e-9           # Posición axial inicial [m]
n_modes_radial = 16 # Modos radiales
m_max = 50           # Modos axiales considerados
N_z = 1000           # Nodos en z (malla refinada)
alpha_grid = 5.0      # Parámetro de densidad de malla
delta = 0.1e-9       # Ancho de transición para D(z)

# =============================================
# 1. Generación de Malla Adaptativa en z
# =============================================
j = np.arange(N_z)
z = z_c * (1 + np.sinh(alpha_grid * j / (N_z - 1)) / np.sinh(alpha_grid))

# =============================================
# 2. Perfil de Coeficiente de Difusión D(z)
# =============================================
def diffusion_coefficient(z):
    S = 0.5 * (1 + np.tanh((z - z_c)/delta))
    return D0 * (np.exp(-lambda_param * z) * (1 - S) + S)

D_values = diffusion_coefficient(z)

# =============================================
# 3. Solución Radial (Funciones de Bessel)
# =============================================
alpha_zeros = jn_zeros(0, n_modes_radial)
alpha_n = alpha_zeros / R
norm_J0_sq = (R**2 / 2) * jv(1, alpha_zeros)**2  # Normas radiales

# =============================================
# 4. Solución Axial con Verificación de Ortogonalidad
# =============================================
eigenvalues_dict = {}
eigenvectors_dict = {}

for n in range(n_modes_radial):
    main_diag = np.zeros(N_z)
    upper_diag = np.zeros(N_z - 1)
    lower_diag = np.zeros(N_z - 1)
    
    # Ensamblaje de la matriz tridiagonal
    for j in range(1, N_z - 1):
        h_j = z[j] - z[j-1]
        h_jp1 = z[j+1] - z[j]
        D_j_plus_half = 0.5 * (D_values[j] + D_values[j+1])
        D_j_minus_half = 0.5 * (D_values[j] + D_values[j-1])
        
        main_diag[j] = D_j_plus_half/h_jp1**2 + D_j_minus_half/h_j**2 + (alpha_n[n]**2)*D_values[j]
        upper_diag[j] = -D_j_plus_half/h_jp1**2
        lower_diag[j-1] = -D_j_minus_half/h_j**2
    
    # Condiciones de frontera
    h_1 = z[1] - z[0]
    main_diag[0] = D_values[0]/h_1**2 + kappa
    upper_diag[0] = -D_values[0]/h_1**2
    main_diag[-1] = 1.0
    lower_diag[-1] = -1.0
    
    # Resolver autovalores
    eigenvalues, eigenvectors = eigh_tridiagonal(main_diag, lower_diag)
    
    # Verificar ortogonalidad de los primeros 10 modos
    for m in range(10):
        for k in range(m+1, 10):
            product = simpson(eigenvectors[:, m] * eigenvectors[:, k] / D_values, x=z)
            if abs(product) > 1e-3:
                print(f"¡Advertencia! Modos {m}-{k} no ortogonales: {product:.2e}")
    
    eigenvalues_dict[n] = eigenvalues
    eigenvectors_dict[n] = eigenvectors

# =============================================
# 5. Cálculo de Coeficientes A_nm (Validación Incluida)
# =============================================
def compute_A_nm(n):
    A_nm_list = []
    for m in range(m_max):
        Z_m_z0 = np.interp(z0, z, eigenvectors_dict[n][:, m])
        norm_Zm_sq = simpson(eigenvectors_dict[n][:, m]**2 / D_values, x=z)
        
        if norm_Zm_sq < 0:
            raise ValueError(f"Norma axial negativa en n={n}, m={m}")
        
        A_nm_val = (jv(0, alpha_n[n] * rho0) * Z_m_z0 / (2 * np.pi * rho0 * norm_J0_sq[n] * norm_Zm_sq))
        A_nm_list.append(A_nm_val)
    
    return A_nm_list

A_nm = Parallel(n_jobs=-1)(delayed(compute_A_nm)(n) for n in range(n_modes_radial))

# =============================================
# 6. Cálculo de G(t) con Estabilización Numérica
# =============================================
def compute_G(t, N_quad=16):
    rho_quad, w_rho = np.polynomial.legendre.leggauss(N_quad)
    rho_quad = 0.5 * R * (rho_quad + 1)
    w_rho *= 0.5 * R
    
    z_quad, w_z = np.polynomial.legendre.leggauss(N_quad)
    z_quad = 0.5 * L * (z_quad + 1)
    w_z *= 0.5 * L
    
    def process_point(i, j):
        rho_i, z_i = rho_quad[i], z_quad[j]
        term = 0.0
        f_i = (3*z_i**2 - rho_i**2) / (rho_i**2 + z_i**2)**(5/2)
        
        # Precomputar J0(α_n ρ_i) para todos los modos radiales
        J0_vals = [jv(0, alpha_n[n] * rho_i) for n in range(n_modes_radial)]
        
        for n in range(n_modes_radial):
            for m in range(m_max):
                Z_m_z = np.interp(z_i, z, eigenvectors_dict[n][:, m])
                decay = np.exp(-(alpha_n[n]**2 * D0 + eigenvalues_dict[n][m]) * t)
                
                # Filtrar términos insignificantes
                if decay > 1e-20:
                    term += A_nm[n][m] * J0_vals[n] * Z_m_z * decay
        
        return w_rho[i] * w_z[j] * f_i * term
    
    results = Parallel(n_jobs=-1)(delayed(process_point)(i, j) for i in range(N_quad) for j in range(N_quad))
    
    return (5/(16*np.pi)) * np.sum(results)

# =============================================
# 7. Ejecución Principal y Generación de Resultados
# =============================================
if __name__ == "__main__":
    # Generar Dz_profiles.pdf
    plt.figure(figsize=(8, 5))
    plt.plot(z*1e9, D_values/D0, 'b-', linewidth=2)
    plt.xlabel('z (nm)', fontsize=12)
    plt.ylabel('D(z)/D₀', fontsize=12)
    plt.title('Perfil del Coeficiente de Difusión', fontsize=14)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.savefig('Dz_profiles.pdf', bbox_inches='tight')
    plt.close()
    
    # Generar eigenvalues.pdf
    plt.figure(figsize=(8, 5))
    for n in range(3):
        eigenvalues = eigenvalues_dict[n][:20]
        plt.semilogy(eigenvalues, 'o-', label=f'Modo Radial n={n}')
    plt.xlabel('Índice del Modo Axial', fontsize=12)
    plt.ylabel('Λₘ² (s⁻¹)', fontsize=12)
    plt.title('Espectro de Autovalores', fontsize=14)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.savefig('eigenvalues.pdf', bbox_inches='tight')
    plt.close()
    
    # Calcular G(t)
    t_values = np.logspace(-9, -3, 50)
    G_values = Parallel(n_jobs=-1)(delayed(compute_G)(t) for t in t_values)
    
    # Forzar no negatividad física
    G_values = np.where(np.array(G_values) < 0, 0, G_values)
    
    # Guardar datos
    np.savetxt('G_vs_t_corregido.dat', np.column_stack((t_values, G_values)),
               header='Tiempo(s) G(t)', fmt='%.3e')
    
    # Gráfico final
    plt.figure(figsize=(8, 5))
    plt.loglog(t_values, G_values, 'ko-', markersize=5, linewidth=1.5)
    plt.xlabel('Tiempo (s)', fontsize=12)
    plt.ylabel('G(t)', fontsize=12)
    plt.title('Función de Correlación nano-NMR', fontsize=14)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.savefig('Gt_parallel.pdf', bbox_inches='tight')
    plt.close()
```
