```python
import glob
import os
import pandas as pd
os.chdir(r"C:\Users\AllanGonzalez\Documents\Repositories\efectos-superficie-nano-RMN\data\Individual_G_vs_t")
# 1) Lista todos los .dat de la carpeta actual
archivos = sorted([f for f in os.listdir('.') if f.endswith('.dat')])

if not archivos:
    raise FileNotFoundError("No se encontraron archivos .dat en el directorio actual.")

# 2) Lee el primer archivo para obtener la columna 'time' y su columna G
primero = archivos[0]
df = pd.read_csv(
    primero,
    comment='#',
    delim_whitespace=True,
    header=None,
    names=['time', f'G_{primero}']
)

# 3) Para cada uno de los demás, solo lee la segunda columna y haz merge por 'time'
for nombre in archivos[1:]:
    df_temp = pd.read_csv(
        nombre,
        comment='#',
        delim_whitespace=True,
        header=None,
        names=['time', f'G_{nombre}']
    )
    df = df.merge(df_temp, on='time')

# 4) Ahora 'df' tiene:
#      ┌───────────┬───────────────┬───────────────┬───────────────┐
#      │   time    │  G_<archivo1> │  G_<archivo2> │  G_<archivo3> │ …
#      └───────────┴───────────────┴───────────────┴───────────────┘

# 5) (Opcional) Muestra las primeras filas para verificar:
print(df.head())

# 6) Guarda todo en un único archivo:
df.to_csv("G_todos_los_kappa.dat", sep=' ', index=False, float_format="%.9e")
print("Archivo combinado guardado como 'G_todos_los_kappa.dat'")
```
