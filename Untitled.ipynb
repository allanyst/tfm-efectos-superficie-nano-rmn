```python
import numpy as np
from scipy.special import jn_zeros
from scipy.optimize import root_scalar
import matplotlib.pyplot as plt
from scipy.special import jv, jn_zeros
from scipy.integrate import quad
from joblib import Parallel, delayed
from scipy.optimize import differential_evolution, dual_annealing

# Parámetros físicos
D1 = 0.5e9  # nm²/s (región superficial)
D2 = 2e9    # nm²/s (región volumétrica)
z_c = 10    # nm (interfaz)
L = 100     # nm (altura total)
kappa = 1   # nm/s (absorción superficial)
R=1000
alpha_n = jn_zeros(0, 3)[0] / R  # Modo radial n=0 (R = 1000 nm)
```

```python
import numpy as np
from scipy.optimize import brentq, differential_evolution, dual_annealing
import matplotlib.pyplot as plt

# # Parámetros físicos globales (ejemplo)
# D1 = 0.5e9   # nm²/s (región superficial)
# D2 = 2e9     # nm²/s (región volumétrica)
# z_c = 10     # nm (posición de la interfaz)
# L = 100      # nm (longitud total)
# kappa = 1    # nm/s (coeficiente de absorción)
# R = 1000     # nm (radio del cilindro)
# alpha_n = 0.0024  # nm⁻¹ (primer modo radial)

def transcendental_eq(Lambda, debug=False):
    """Función trascendental mejorada con manejo de casos especiales."""
    try:
        # Prevenir valores negativos en las raíces cuadradas
        beta_1_sq = (Lambda**2 - alpha_n**2 * D1) / D1
        beta_2_sq = (Lambda**2 - alpha_n**2 * D2) / D2
        
        if beta_1_sq < 0 or beta_2_sq < 0:
            if debug: print(f"Λ={Lambda:.2e}: Raíz cuadrada negativa (β1²={beta_1_sq:.2e}, β2²={beta_2_sq:.2e})")
            return np.inf
            
        beta_1 = np.sqrt(beta_1_sq + 1e-12)
        beta_2 = np.sqrt(beta_2_sq + 1e-12)
        
        # Término crítico con la tangente
        arg_tan = beta_2*(L - z_c)
        epsilon = 1e-12  # Para evitar división por cero
        
        # Reescribir la ecuación sin tan() explícito
        numerator = (D1 * beta_1 * np.cos(arg_tan) - (kappa * D2 * beta_2)/(D1 * beta_1 + epsilon) * np.sin(arg_tan)) * np.sin(beta_1 * z_c)
        denominator = (kappa * beta_1 * np.cos(arg_tan) + D2 * beta_2 * np.sin(arg_tan)) * np.cos(beta_1 * z_c)
        
        result = numerator + denominator
        
        if debug: 
            print(f"Λ={Lambda:.2e}, Resultado={result:.2e}")
            print(f"β1={beta_1:.2e}, β2={beta_2:.2e}")
            
        return result
        
    except Exception as e:
        print(f"Error en Λ={Lambda:.2e}: {str(e)}")
        return np.inf

# Función wrapper para métodos de optimización
def objective_function(Lambda):
    return np.abs(transcendental_eq(Lambda[0]))

# 1. Método de Brentq (requiere intervalo con cambio de signo)
def test_brentq(L_min=1e3, L_max=1e5):
    try:
        # Verificar cambio de signo
        f_min = transcendental_eq(L_min)
        f_max = transcendental_eq(L_max)
        
        if f_min * f_max > 0:
            raise ValueError("No hay cambio de signo en el intervalo")
            
        root = brentq(transcendental_eq, L_min, L_max, xtol=1e-6)
        print(f"\nBrentq convergió: Λ = {root:}")
        print(f"Residuo: {transcendental_eq(root):}")
        return root
        
    except Exception as e:
        print(f"\nError en Brentq: {str(e)}")
        return None

# 2. Differential Evolution
def test_diff_evolution(L_min=1e3, L_max=1e7):
    bounds = [(L_min, L_max)]
    
    result = differential_evolution(
        objective_function,
        bounds,
        strategy='best1bin',
        popsize=15,
        mutation=(0.5, 1.0),
        recombination=0.7,
        tol=1e-6,
        maxiter=1000,
        seed=42,
        disp=True
    )
    
    if result.success:
        print(f"\nDifferential Evolution convergió: Λ = {result.x[0]}")
        print(f"Residuo: {result.fun:}")
        return result.x[0]
    else:
        print("\nDifferential Evolution falló")
        return None

# 3. Dual Annealing
def test_dual_annealing(L_min=1e3, L_max=1e7):
    bounds = [(L_min, L_max)]
    
    result = dual_annealing(
        objective_function,
        bounds,
        maxiter=1000,
        seed=42,
        no_local_search=False,
        initial_temp=1e4,
        visit=2.6,
        accept=-5.0,
    )
    
    if result.success:
        print(f"\nDual Annealing convergió: Λ = {result.x[0]:}")
        print(f"Residuo: {result.fun:}")
        return result.x[0]
    else:
        print("\nDual Annealing falló")
        return None

# Función para análisis gráfico
def plot_transcendental(L_min=1e3, L_max=1e5, num_points=1000):
    Lambdas = np.linspace(L_min, L_max, num_points)
    values = [transcendental_eq(L) for L in Lambdas]
    
    plt.figure(figsize=(10, 6))
    plt.plot(Lambdas, values, label='f(Λ)')
    plt.axhline(0, color='red', linestyle='--')
    plt.xlabel('Λ (s⁻¹)')
    plt.ylabel('Valor de la ecuación trascendental')
    plt.title('Comportamiento de la ecuación trascendental')
    plt.yscale('symlog', linthresh=1e-6)
    plt.grid(True)
    plt.legend()
    plt.show()

# Uso
if __name__ == "__main__":
    # Análisis gráfico primero
    plot_transcendental(L_min=alpha_n*np.sqrt(D2), L_max=2e4)
    
    # Prueba de métodos
    print("\n=== Probando Brentq ===")
    lambda_brentq = test_brentq(alpha_n*np.sqrt(D2), 2e4)
    
    print("\n=== Probando Differential Evolution ===")
    lambda_de = test_diff_evolution(alpha_n*np.sqrt(D2), 2e4)
    
    print("\n=== Probando Dual Annealing ===")
    lambda_da = test_dual_annealing(alpha_n*np.sqrt(D2), 2e4)
    
    # Validación cruzada
    if lambda_brentq and lambda_de and lambda_da:
        print("\nValidación cruzada:")
        print(f"Brentq vs DE: {abs(lambda_brentq - lambda_de)/lambda_brentq*100:}% diferencia")
        print(f"Brentq vs DA: {abs(lambda_brentq - lambda_da)/lambda_brentq*100:}% diferencia")
```
